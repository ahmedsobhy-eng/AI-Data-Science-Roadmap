```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

# 1. نجيب الداتا من الملف الأصلي تاني
df = pd.read_csv("players_data-2026_2027.csv")

# 2. الحل السحري: أي خانة فاضية حط مكانها صفر بدل ما تمسح اللاعب
df.fillna(0, inplace=True)

# 3. نعزل الأرقام، ونفصل المعطيات عن الأهداف
numeric_df = df.select_dtypes(include=['number'])
X = numeric_df.drop("Gls", axis=1)
y = df["Gls"]

# 4. نقسم الداتا (دلوقتي الأداة هتلاقي لعيبة بجد تقسمها!)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 5. نوحد المقاسات
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 6. ندرب الموديل
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# 7. التوقع
predictions = model.predict(X_test_scaled)

print("تم التدريب والتوقع بنجاح! 🎉")
print("عدد اللعيبة اللي اتدربنا عليهم:", len(X_train))
```

```python
# بنعمل جدول يعرض الأهداف الحقيقية جنب توقع الموديل
results = pd.DataFrame({
    'Real_Goals': y_test,
    'Predicted_Goals': predictions.round(1) # بنقرب الرقم عشان ميبقاش فيه كسور كتير
})

# نعرض أول 10 لعيبة في الامتحان
print(results.head(10))

# نحسب الموديل بيغلط في كام جول في المتوسط (بنجيب أداة تحسب نسبة الخطأ)
from sklearn.metrics import mean_absolute_error
error = mean_absolute_error(y_test, predictions)

print(f"\nAverage Error: الموديل بيغلط في حوالي {error:.1f} أهداف لكل لاعب")
```
